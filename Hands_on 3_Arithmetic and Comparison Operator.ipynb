# COMPARISON AND ARITHMETIC OPERATORS

## Number One: Employee payroll and Bonus Eligibility

```python
# The code aggregate an employee's total bonus to determine his eligibility criteria
basic_salary = 180000
performance_bonus = 45000

total_salary = basic_salary + performance_bonus
print("#",total_salary)

print("Eligibility for premium benefit:", total_salary >= 220000)
```

## Number Two: Student Performance Evaluation

```python
# This code assesses a students's score to determine if he passed his assessments
python = 72
sql = 65
excel = 80
power_bi = 75

total_score = python + sql + excel + power_bi

average_score = total_score / 4
print("Average score:", average_score)

print("Passed:", average_score >= 70)

```

## Number Three: Inventory Management System

```python
# This code evaluates a warehouse stock to determine when to trigger reorder inventory.
initial_stock = 250
items_sold = 85

remaining_stock = initial_stock - items_sold
print("Remaining stock:", remaining_stock, "Units")
print("Reorder needed:", remaining_stock < 200)


```

## Number Four: Banking Withdrawal Checker

```python
# This code assess a customer's balance to know if his withdrawal will be approved
account_balance = 520000
withdrawal = 150000

remaining_balance = account_balance - withdrawal

print("Remaining Balance: #",remaining_balance)
print("Withdrawal approved",account_balance >= withdrawal)
```

## BONUS CHALLENGE

```python
# A smart shopping discount calculator
laptop_bag = 18500
wireless_mouse = 12000
keyboard = 24500
usb_drive = 5000
discount = 0.10

total_purchase = laptop_bag + wireless_mouse + keyboard + usb_drive
print("Total Purchase: #", total_purchase)

print(total_purchase >= 50000)

discount_amount = total_purchase * discount
print("Discount amount: #", discount_amount )

sales_price = total_purchase - discount_amount 
print("Final_Sales_Price: #", sales_price)
```
