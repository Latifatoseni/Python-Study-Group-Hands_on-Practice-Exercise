## Study Buddy Bonus Challenge


```python
# The code determines if a customer is eligible to place an order

logged_in = True
item_instock = True
payment_method = True
wallet_balance = False
account_suspended = False

print(logged_in and item_instock and (payment_method or wallet_balance) and account_suspended)
print ("Order Failed")
```

## Scenario 1

```python
logged_in = True
item_instock = True
payment_method = True
wallet_balance = False
account_suspended = False

print((not logged_in or item_instock) and (payment_method or wallet_balance) and account_suspended)
print ("Order Failed")

#The item was not logged in but it was in stock; the Logical operator is still "and", joining the other or statement.
#The first statement will return true, the second will return true but because there is and joining the two parenthesis 
# The account suspension was also false hence we have "True and True and False which will result to false", hence the result.
```

## Scenario 2

```python
logged_in = True
item_instock = True
payment_method = True
wallet_balance = False
account_suspended = False

print((not logged_in or item_instock) or (payment_method or wallet_balance) or account_suspended)
print ("Order confirmed")

# The logical operator between the parenthesis was converted form and to "or".
# The first parentheis will return true, the second will return true.
# The result will then be "True or True or False", which will return true.
```

## Scenario 3

```python
logged_in = True
item_instock = True
payment_method = True
wallet_balance = False
account_suspended = False

print((logged_in and not item_instock) and (not payment_method or not wallet_balance) and not account_suspended)
print ("Order Failed")

# The first parethesis result returns False since one is false
# The second parenthesis result returns True since one is true
# The result becomes " False and True and True" which in turn returns False
```
