# Python Assignment Operators

## Student Score Tracker

```python
# Number one 
# A python code that updates student code
score = int(input("Enter your score"))
score += 15
score -= 5
score *= score
print("The Final score is",score)
```

## Digital Wallet simulator


```python
# Number Two
# A python code that updates customer accounts after every transaction
balance = 25000
balance += int(input("Enter your deposit"))
balance -= int(input("Enter your withdrawal"))
balance /= 3
print("The final balance is", balance)
print("The data type:", type(balance))
```

## Inventory management project

```python
# Number Three
# A python code that updates customer's items and stock arrival
stock = int(input("What was the initial stock"))
stock -= int(input("How many items were sold"))
print("The remaining stock after sales is:", stock)

stock += int(input("How many new items arrived"))
print("The updated stock after restocking is:", stock)

stock //= int(input("How many items fit into one carton"))
print("The number of full cartons is", stock)

updated_restocking = 197
carton_items = 12
updated_restocking %= carton_items
print("The loose items are:", updated_restocking)
```
