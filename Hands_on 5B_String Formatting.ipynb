## Student Report Card

```python
# Number one
# Build a student report for a secondary school

student_name = "Esther Johnson"
age = 17
subject = "Mathematics"
score = 91.75
print("=======================================================")
print("               STUDENT REPORT                          ")
print("=======================================================")
print(f" Student: {student_name}")
print(f" Age:{age} years")
print(f" Subject: {subject}")
print(f" Score: {score}% \n")

print(f""" Congratulations Esther Johnson!
 You passed Mathematics.""")
print("=======================================================")
```

## Payroll System

```python
# Number Two
# Generate automatic payroll receipt

employee_name = "David Musa"
department = "Finance"
basic_salary = 185000
bonus = 35000
tax_rate = 0.075  #7.5%

gross_salary = basic_salary + bonus 
tax_amount = gross_salary * tax_rate
net_salary = gross_salary - tax_amount

print("===========================================")
print("             PAYROLL AMOUNT                ")
print("===========================================")
print(f" Employee: {employee_name}")
print(f" Department: {department}")
print(f" Basic Salary: ₦{basic_salary:.2f}")
print(f" Bonus: ₦{bonus:.2f}")
print(f" Gross Salary: ₦{gross_salary:.2f}")
print(f" Tax (7.5%): ₦{tax_amount:.2f}")
print(f" Net Salary: ₦{net_salary:.2f}")
print("===========================================")



```

## E- Commerce Order Receipt

```python
# Number Three
# Display customer order receipt

customer_name = "Aminat Bello"
product = "Wireless Headphones"
quantity = 3
unit_price = 27500 #  ₦
discount = 0.1     # 10%

total_cost = quantity * unit_price
discounted_amount = total_cost * discount
final_amount = total_cost - discounted_amount

print("===========================================")
print("             ONLINE STORE RECEIPT          ")
print("===========================================")
print(f" Customer: {customer_name}")
print(f" Product: {product}")
print(f" Quantity: {quantity}")
print(f" Unit Price: ₦{unit_price:.2f}")
print(f" Total Cost: ₦{total_cost:.2f} ")
print(f" Discount Amount : ₦{discounted_amount:.2f} ")
print(f" Final Amount: ₦{final_amount:.2f} \n")

print(f" Thank you for shopping with us!")
print("===========================================")


```
